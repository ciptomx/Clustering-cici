```python
# ==========================================
# 1. IMPORT LIBRARY
# ==========================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

from neuralforecast import NeuralForecast
from neuralforecast.models import PatchTST
from neuralforecast.losses.numpy import mae, mse, rmse

# ==========================================
# 2. UPLOAD FILE DATASET
# ==========================================
print("\n[INFO] Silakan upload file dataset (DataAWSFinal.xlsx)...")
uploaded = files.upload()
file_path = 'DataAWSFinal.xlsx'

# ==========================================
# 3. PARAMETER MODEL & EVALUASI
# ==========================================
WINDOW_SIZE = 96   # Input 4 hari ke belakang
TEST_SIZE = 720
HORIZON = TEST_SIZE # Prediksi langsung sepanjang periode test
MAX_STEPS = 5000

# ==========================================
# 4. LOAD & PREPROCESSING
# ==========================================
print("\n[INFO] Memuat dan memproses data...")
try:
    df = pd.read_excel(file_path)
except:
    # Antisipasi jika nama file berbeda saat diupload
    import os
    file_path = list(uploaded.keys())[0]
    df = pd.read_excel(file_path)

# Rename sesuai standar NeuralForecast
df_nf = df.rename(columns={'waktu': 'ds', 'waterlevel': 'y'})
df_nf['unique_id'] = 'AWS_Station'
df_nf['ds'] = pd.to_datetime(df_nf['ds'])
df_nf = df_nf[['unique_id', 'ds', 'y']].sort_values('ds').reset_index(drop=True)

# ---------------------------------------------------------
# PERBAIKAN: SPLIT DATA DULU, BARU INTERPOLASI (Saran Dosen)
# ---------------------------------------------------------
train_size = len(df_nf) - TEST_SIZE
df_train = df_nf.iloc[:train_size].copy()
df_test  = df_nf.iloc[train_size:].copy()

# Hanya lakukan interpolasi pada data TRAIN agar tidak bocor ke TEST
df_train['y'] = df_train['y'].interpolate(method='linear').ffill().bfill()

print(f"Total Data  : {len(df_nf)}")
print(f"Data Train  : {len(df_train)}")
print(f"Data Test   : {len(df_test)}")

# ==========================================
# 5. MEMBANGUN MODEL (PERBAIKAN MAX_STEPS)
# ==========================================
print("\n[INFO] Memulai Training Model PatchTST (Estimasi 5000 steps)...")

models = [
    PatchTST(
        h=HORIZON,
        input_size=WINDOW_SIZE,
        max_steps=MAX_STEPS,  # Menggunakan max_steps sesuai error sebelumnya
        batch_size=32,
        encoder_layers=3,
        learning_rate=1e-3,
        scaler_type='standard'
    )
]

nf = NeuralForecast(
    models=models,
    freq='h' # Menggunakan lowercase 'h'
)

# Training model menggunakan data yang sudah bersih
nf.fit(df=df_train)
print("[INFO] Training selesai.")

# ==========================================
# 6. PREDIKSI (PERBAIKAN DATA LEAKAGE)
# ==========================================
print("\n[INFO] Melakukan Prediksi...")

# PERBAIKAN: Prediksi dilakukan HANYA berdasarkan df_train.
# Model akan memproyeksikan masa depan tanpa melihat df_test.
forecast_df = nf.predict(df=df_train)
forecast_df = forecast_df.reset_index()

# Gabungkan hasil prediksi dengan data test asli untuk evaluasi
result_df = pd.merge(
    df_test,
    forecast_df[['ds', 'PatchTST']],
    on='ds',
    how='inner'
)

# Jika ada nilai kosong di y aktual pada data test, kita hapus agar evaluasi valid
result_df = result_df.dropna(subset=['y'])

# ==========================================
# 7. EVALUASI MODEL
# ==========================================
y_true = result_df['y'].values
y_pred = result_df['PatchTST'].values

mae_val = mae(y_true, y_pred)
mse_val = mse(y_true, y_pred)
rmse_val = rmse(y_true, y_pred)

print("\n" + "="*40)
print("  HASIL EVALUASI MODEL ( PatchTST )")
print("="*40)
print(f"MAE  : {mae_val:.5f}")
print(f"MSE  : {mse_val:.5f}")
print(f"RMSE : {rmse_val:.5f}")
print("="*40)

# ==========================================
# 8. VISUALISASI
# ==========================================
plt.figure(figsize=(15, 6))

# Kita plot 168 jam pertama (1 minggu) dari periode test agar terlihat jelas
plot_df = result_df.head(168)

plt.plot(plot_df['ds'], plot_df['y'], label='Aktual (Raw Data)', color='black', linewidth=2)
plt.plot(plot_df['ds'], plot_df['PatchTST'], label='Prediksi PatchTST', color='red', linestyle='--')

plt.title(f'Evaluasi Prediksi Waterlevel (1 Minggu Pertama Test Set)\nRMSE: {rmse_val:.4f}', fontsize=14)
plt.xlabel('Waktu')
plt.ylabel('Water Level')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Tampilkan 24 jam pertama hasil prediksi dalam bentuk tabel
print("\nDetail 24 Jam Pertama Prediksi:")
print(result_df[['ds', 'y', 'PatchTST']].head(24))
```
