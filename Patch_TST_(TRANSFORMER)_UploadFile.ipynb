```python
# ==========================================
# 1. IMPORT LIBRARY
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Upload file dataset ke notebook
from google.colab import files

# Import Library NeuralForecast
from neuralforecast import NeuralForecast
from neuralforecast.models import PatchTST
from neuralforecast.losses.numpy import mae, mse, rmse

# ==========================================
# 2. UPLOAD FILE DATASET (.xlsx)
# ==========================================

print("\n[INFO] Silakan upload file dataset (.xlsx)...")

uploaded = files.upload()

# Pastikan nama file sama dengan yang di-upload
file_path = 'DataAWSFinal.xlsx'

# ==========================================
# 3. PARAMETER MODEL
# ==========================================

WINDOW_SIZE = 96   # Input window (96 jam)
HORIZON = 24       # Horizon prediksi (24 jam)
EPOCHS = 100       # Jumlah epoch

# ==========================================
# 4. LOAD DATA
# ==========================================

print("\n[INFO] Memuat data...")

try:
    df = pd.read_excel(file_path)
except FileNotFoundError:
    raise FileNotFoundError(
        "File tidak ditemukan! Pastikan nama file benar."
    )

# ==========================================
# 5. PREPROCESSING DATA
# ==========================================

target_col = 'waterlevel'
date_col = 'waktu'

# Rename kolom
df_nf = df.rename(columns={
    date_col: 'ds',
    target_col: 'y'
})

# Tambahkan ID
df_nf['unique_id'] = 'AWS_Station'

# Konversi waktu
df_nf['ds'] = pd.to_datetime(df_nf['ds'])

# Urutkan data
df_nf = df_nf[['unique_id', 'ds', 'y']]\
.sort_values('ds')\
.reset_index(drop=True)

# Handling Missing Value
df_nf['y'] = df_nf['y']\
.interpolate(method='linear')\
.ffill()\
.bfill()

print("[INFO] Preprocessing selesai.")

# ==========================================
# 6. SPLIT DATA TRAIN & TEST
# ==========================================

test_size = HORIZON
train_size = len(df_nf) - test_size

df_train = df_nf.iloc[:train_size]
df_test  = df_nf.iloc[train_size:]

print("\n[INFO] Data Split selesai.")
print("Data Train:", len(df_train))
print("Data Test :", len(df_test))

# ==========================================
# 7. MEMBANGUN MODEL PatchTST
# ==========================================

print("\n[INFO] Memulai Training Model PatchTST...")

models = [
    PatchTST(
        h=HORIZON,
        input_size=WINDOW_SIZE,
        max_steps=EPOCHS,
        encoder_layers=3,
        learning_rate=1e-3,
        scaler_type='standard'
    )
]

nf = NeuralForecast(
    models=models,
    freq='H'
)

# Training model
nf.fit(df=df_train)

print("[INFO] Training selesai.")

# ==========================================
# 8. PREDIKSI
# ==========================================

print("\n[INFO] Melakukan Prediksi...")

forecast_df = nf.predict(futr_df=df_test)

# Gabungkan hasil
forecast_df = forecast_df.reset_index()
actual_df = df_test.reset_index(drop=True)

result_df = pd.merge(
    actual_df,
    forecast_df,
    on=['ds', 'unique_id'],
    how='inner'
)

# ==========================================
# 9. EVALUASI MODEL
# ==========================================

y_true = result_df['y'].values
y_pred = result_df['PatchTST'].values

mae_val = mae(y_true, y_pred)
mse_val = mse(y_true, y_pred)
rmse_val = rmse(y_true, y_pred)

print("\n" + "="*40)
print("  HASIL EVALUASI MODEL (PatchTST)")
print("="*40)
print(f"MAE  : {mae_val:.5f}")
print(f"MSE  : {mse_val:.5f}")
print(f"RMSE : {rmse_val:.5f}")
print("="*40)

# ==========================================
# 10. VISUALISASI HASIL
# ==========================================

plt.figure(figsize=(14, 6))

plt.plot(
    result_df['ds'],
    result_df['y'],
    label='Aktual',
    color='black',
    linewidth=2
)

plt.plot(
    result_df['ds'],
    result_df['PatchTST'],
    label='Prediksi PatchTST',
    color='red',
    linestyle='--',
    marker='o'
)

plt.title(
    f'Prediksi Waterlevel (PatchTST)\nRMSE: {rmse_val:.4f}',
    fontsize=14
)

plt.xlabel('Waktu')
plt.ylabel('Water Level')

plt.legend()
plt.grid(True, alpha=0.3)

plt.show()

# ==========================================
# 11. TAMPILKAN HASIL 24 JAM
# ==========================================

print("\nDetail Hasil Prediksi (24 Jam):")

print(
    result_df[
        ['ds', 'y', 'PatchTST']
    ].head(24)
)
```
